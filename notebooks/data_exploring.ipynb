```python
from pyspark.sql.functions import col, count, when

valeurs_nulles_df = df.select([count(when(col(c).isNull(), c)).alias(c) for c in df.columns])

display(valeurs_nulles_df)
```

```python
from pyspark.sql.functions import isnan

colonnes_exclues = ["Label", "ingestion_timestamp", "source_file"]

exprs = []
for c in df.columns:
    if c == "Label":
        exprs.append(count(when((col(c) == ""), c)).alias(c))
    
    elif c not in colonnes_exclues:
        exprs.append(count(when(isnan(col(c)), c)).alias(c))
        
valeurs_manquantes_df = df.select(exprs)
display(valeurs_manquantes_df)
```

```python
total_lignes = df.count()
lignes_uniques = df.dropDuplicates().count()
nb_doublons = total_lignes - lignes_uniques

print(f"Nombre total de lignes : {total_lignes}")
print(f"Nombre de lignes uniques : {lignes_uniques}")
print(f"Nombre de doublons : {nb_doublons}")
```

```python
from pyspark.sql.functions import col

df = spark.table("xaids_catalogue.`01_bronze`.cicds2017_raw_full")

diff_count = df.filter(
    col("Fwd_Header_Length34") != col("Fwd_Header_Length55")
).count()
print(f"Lignes où les 2 colonnes dupliquées diffèrent : {diff_count}")
```

```python
from pyspark.sql import functions as F

range_cols = ["Fwd_Header_Length34", "Bwd_Header_Length", "min_seg_size_forward"]

for c in range_cols:
    print(f"\n=== {c} (valeurs positives uniquement) ===")
    
    positive_df = df.filter(F.col(c) > 0)
    
    quantiles = positive_df.approxQuantile(
        c, [0.01, 0.25, 0.5, 0.75, 0.9, 0.95, 0.99, 0.999], relativeError=0.001
    )
    print(f"Percentiles [1%, 25%, 50%, 75%, 90%, 95%, 99%, 99.9%] : {quantiles}")
    
    positive_df.select(
        F.min(c).alias("min"),
        F.max(c).alias("max"),
        F.mean(c).alias("mean"),
        F.stddev(c).alias("stddev"),
        F.count(c).alias("count")
    ).show()

    above_threshold = positive_df.filter(F.col(c) >= 100000).count()
    total_positive = positive_df.count()
    print(f"Lignes >= 100000 : {above_threshold} / {total_positive} "
          f"({above_threshold/total_positive:.4%})")
```
